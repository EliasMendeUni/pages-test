A short chain: process, points, structure factor.

```python
import numpy as np
from pagestest import Pointprocess

points = Pointprocess(box=4.0, spacing=1.0).rect_lattice()
points.n, points.dmean
```

```python
g = 2 * np.pi
points.structure_factor(np.array([[0.0, 0.0], [g, 0.0], [g / 2, 0.0]])).round(6)
```
